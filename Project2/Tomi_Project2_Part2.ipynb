# Data Wrangling
### Adding a new column based on another

```python
hi_new=human_interaction.copy()
hi_new
```

```python
def drained(Time_spent_Alone):
    if pd.isnull(Time_spent_Alone):
        return 'N/A'
    elif Time_spent_Alone < 5:
        return 1
    else:
        return 2

def fright(Friends_circle_size):
    if pd.isnull(Friends_circle_size):
        return 'N/A'
    elif Friends_circle_size <= 5:
        return 2
    else:
        return 1
```

```python
hi_new['Drained_after_events']=hi_new['Time_spent_Alone'].apply(drained)
hi_new['Stage_fright']=hi_new['Friends_circle_size'].apply(fright)
hi_new
```

```python
hi_new.head(11)
```

###### The new column Drained after events is  based on the person's rating of how much time they spend alone. If they spend less than 5 hours alone, they're considered extroverts who rarely feel drained after socializing. So those who are considered extroverts are given the number 1, in case a correlation table is needed later on. And if they have 5 or less friends they are consdered introverts, so they are given the number 2.

## **Correlation Heat Map of Numeric Features**
#### The heat map utilizes a color-coded scale to visualize the strength and direction of relationships between various behaviors in the dataset. In this specific visualization, darker shades indicate a strong positive relationship, where two factors increase together, while lighter colors represent values near zero, suggesting little to no connection. A key trend revealed by this map is the very strong negative correlation of -0.718 between time spent alone and social event attendance, meaning that as an individual's solitary time increases, their participation in social gatherings significantly decreases. This is a critical finding because it demonstrates that social behaviors are not random occurrences; rather, they are deeply and predictably tied to the specific amount of solitary time a person requires to function.

```python
sns.boxplot(hi_new, x='Drained_after_socializing', y='Time_spent_Alone', color='brown')
plt.title('Hours spent alone vs.State after outings')
plt.savefig('Project2_Graphs/Alone_time_vs._State_after_outings.jpg')
plt.show()
```

```python
fig, axes=plt.subplots(1,2, figsize=(12,6))
sns.barplot(data=hi_new, x='Drained_after_socializing', y='Social_event_attendance', hue='Personality',palette='dark:brown',ax=axes[0])
axes[0].set_title('Frequency of social outing by State after outings ')
axes[0].set_xlabel('Drained after socializing')
axes[0].set_ylabel('Social outing')
sns.barplot(data=hi_new, x='Drained_after_socializing', y='Going_outside', hue='Personality',palette='dark:brown',ax=axes[1])
axes[1].set_title('Frequency of Outside appearence by State after outings')
axes[1].set_xlabel('Drained after socializing')
axes[1].set_ylabel('Going outside')
plt.savefig('Project2_Graphs/subplots.jpg')
plt.show()
```

#### These side-by-side bar charts provide a comparison that reveals how feeling drained affects different types of movement beyond just parties. The first plot shows that both Introverts and Extroverts significantly reduce their social outings when they feel exhausted. The second plot, ‘Frequency of Outside Appearance,’ reveals a broader trend: feeling drained doesn't just impact social events; it reduces general activity, such as running errands or just being outdoors. Across both groups, those who are not drained go outside roughly four to five times more often than those who are, suggesting that "social drainage" acts as a major barrier to nearly all forms of movement.

```python
state_after_outings = hi_new.groupby('Drained_after_socializing', observed=True )['Going_outside'].mean() * 100
state_after_outings.plot(kind='bar', color='brown')
plt.title('Frequency of social outing by State after outings')
plt.xlabel('Drained after socializing')
plt.ylabel('Going outside')
plt.xticks(rotation=0)
plt.savefig('Project2_Graphs/group_drained_after_socializing by_the_by_the_going_outside_avg.jpg')
plt.show()
```

#### This Bar Chart summarizes the overall frequency of going outside based specifically on a person's state after a social outing. By looking at the average values, it becomes clear that there is a stark divide; the group that feels drained has a much lower overall spread of activity, reaching only a fraction of the height of the non-drained group. This visualization reveals that the state of being drained is perhaps the most influential factor in the entire dataset, as it consistently predicts lower social circles, more time spent alone, and a drastic reduction in how often a person interacts with the world outside their home.

# **Summary and Conclusion**
### The data reveals a clear and consistent divide in social behavior driven primarily by personality types and energy management. We observed that the majority of the 2,900 individuals surveyed maintain small, intimate friend groups of one to six people, while a small subset of highly social individuals maintains much larger circles. A central discovery in this analysis is the negative relationship between solitude (alone time) and social engagement. As the time spent alone increases, the frequency of social outings and the size of friend groups decrease predictably. Furthermore, the data highlights that social drainage is a universal factor; regardless of whether someone identifies as an Introvert or an Extrovert, feeling drained after interaction leads to a significant reduction in both social attendance and general outdoor activity.
### In conclusion, this study demonstrates that human interaction is not a random occurrence but a balanced system of energy and preference. The data shows that behavior is shaped by a continuous feedback loop between social environments and biological factors. While an individual’s personality provides a biological baseline for how they process social stimulation, their state after outings acts as a gatekeeper for future movement. Those who feel biologically recharged by others are far more likely to engage with the world, while those who feel drained require substantial solitary time, often up to eight hours, to achieve homeostasis and recover. Ultimately, these visualizations prove that while environments provide the setting for interaction, our need for energy management is what truly dictates the rhythm of our social lives.
