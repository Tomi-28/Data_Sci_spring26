# Filtering Data

```python
df_new=df.copy()
df_new
```

```python
df_new['Age']
```

```python
#df_new['Total']=df_new['Math']+df_new['English']+df_new['Chemistry']+df_new['Biology']+df_new['Physics']
df_new['Total']=df_new[['Math','English','Chemistry','Biology','Physics']].sum(axis=1)
```

```python
df_new
```

```python
df_new['Sci_total']=df_new[['Chemistry','Biology','Physics']].sum(axis=1)
df_new['Percentage']=(df_new['Total']/500)*100
df_new
```

```python
def grade(percentage):
    if percentage >=90:
        return 'A'
    elif percentage >=80:
        return 'B'
    elif percentage >=70:
        return 'C'
    elif percentage >=60:
        return 'D'
    else:
        return 'F'
```

```python
df_new['Grade']=df_new['Percentage'].apply(grade)
df_new
```

```python
df_new['Percentage_bonus']=df_new['Percentage'].apply(lambda x:x+5)
df_new
```

```python
print('Data Frame before dropping bonus column:\n', df_new)
df_new.drop(columns=['Percentage_bonus'],inplace=True)
print('Data Frame after dropping bonus column:\n', df_new)
```

```python
df_new
```
