```python
import numpy as np
import pandas as pd
```

```python
n=14
np.random.seed(45)
name=['Jordyn','Alex','Kasey','Matthew','Lili','Sophia','Debby','Howard','Jace','Rosy','Jasmin','Toni','Robby','Maddy']
hrs=np.random.randint(8,20,n)
credit=np.random.randint(12,21,n)
school=np.random.choice(['SOB','COAS','SOC','CNAHS','CEA'],n)
student_data=pd.DataFrame(hrs,columns=['Weekly_Hrs_Spent_Studying'])
student_data
```

```python
student_data.insert(0,'#_of_Credit',credit)
```

```python
student_data.insert(0,'School',school)
```

```python
student_data.insert(0,'Name',name)
```

```python
student_data
```

```python
#add an extra column not listed in the table
def majors(name):
    if name =='Jordyn':
        return 'Finance'
    elif name =='Alex':
        return 'Physics'
    elif name =='Kasey':
        return 'Nursing'
    elif name =='Matthew':
        return 'Physical Therapy'
    elif name =='Lili':
        return 'Advertising'
    elif name =='Sophia':
        return 'Marketing'
    elif name =='Debby':
        return 'Chemical Engineering'
    elif name =='Howard':
        return 'Biology'
    elif name =='Jace':
        return 'Finance'
    elif name =='Rosy':
        return 'Journalism and Media'
    elif name =='Jasmin':
        return 'Chemistry'
    elif name =='Toni':
        return 'Journalism'
    elif name =='Robby':
        return 'Nursing'
    else:
        return 'Business'

student_data['Major']=student_data['Name'].apply(majors)
```

```python
student_data
```

```python
#save DF as a CSV file
student_data.to_csv('student_data_collected.csv',index=False)
```
