# **Conceptual Questions**
### Define Data Science
- Data Science is the process of collecting and interpreting raw structured and unstructured data to predict trends and make decisions.
### Components of Data Science
- There are 2 main components of Data Science:
	- ***Qualitative***: descriptive information that is divided into 3 types:
    	- ***Binomial data***- data with only two options (e.g. male or female, true or false)
        - ***Ordinal data***- data that's in correct order (e.g. breakfast, lunch, dinner)
        - ***Unordered data***- data that's in no specific order (e.g. girls, class, red)
    - ***Quantitative***: numerical information that is divided into 2 types:
    	- ***Discrete data***- data that's countable (e.g. number of cars, number of students)
        - ***Continuous data***- data that's measurable (e.g. height, length), and has 2 types;
        	- **interval**- the data is rational at 0, e.g. temperature
            - **ratio**- the data is irrational at 0, e.g. length
### Hashtag symbol
- The hashtag symbol in Python is used to make a comment
### Difference between a string and a number with example
- A string is treated like a word, e.g., '12' is read as a word and when calculated with '10', the result is 1210, whereas a number is treated like a number, e.g., 12 is read as a value, and when calculated with 10, the result is 22.
### Type Conversion
- Python requires type conversion for user input because the built-in input() function returns the data as a string by default.

## Write a Python program to
- ask user to enter the # of courses you're taking 
- ask user to enter the credits per course
- calculates the total # of credits
- print the total clearly

```python
course_num=int(input('enter the # of courses you are taking this semseter:'))
credits_list=[]
for i in range(1, course_num + 1):
    credits=float(input(f'enter the # of credits for course {i}:'))
    credits_list.append(credits)
    print(i)
total_credits=sum(credits_list)
print(f'The total number of credits I am taking this semester is: {total_credits}')
```
